```python
def remove_footer_from_pdf(input_path, output_path, footer_height=60):
    """
    Remove footer from PDF by cropping the bottom portion of each page
    
    Parameters:
    input_path (str): Path to input PDF
    output_path (str): Path for output PDF
    footer_height (int): Height in pixels to crop from bottom (default: 60)
    """
    try:
        # Open the PDF
        doc = fitz.open(input_path)
        print(f"Opened PDF with {doc.page_count} pages")
        
        skipped_pages = []
        
        # Process each page
        for page_num in range(doc.page_count):
            page = doc[page_num]
            rect = page.rect
            
            # Check if footer_height is reasonable for this page
            if footer_height >= rect.height:
                print(f"⚠️ Skipping page {page_num + 1}: footer height ({footer_height}) >= page height ({rect.height})")
                skipped_pages.append(page_num + 1)
                continue
            
            # Create new rectangle excluding footer area
            new_rect = fitz.Rect(0, 0, rect.width, rect.height - footer_height)
            
            # Validate the crop box is within the media box
            try:
                page.set_cropbox(new_rect)
            except Exception as crop_error:
                print(f"⚠️ Skipping page {page_num + 1}: {str(crop_error)}")
                skipped_pages.append(page_num + 1)
                continue
            
            if (page_num + 1) % 100 == 0:  # Progress indicator every 100 pages
                print(f"Processed {page_num + 1} pages...")
        
        # Save the modified PDF
        doc.save(output_path)
        doc.close()
        
        print(f"✅ Successfully saved cleaned PDF to: {output_path}")
        print(f"Cropped {footer_height} pixels from bottom of each page")
        
        if skipped_pages:
            print(f"⚠️ Skipped {len(skipped_pages)} pages due to dimension issues: {skipped_pages[:10]}{'...' if len(skipped_pages) > 10 else ''}")
        
    except Exception as e:
        print(f"❌ Error processing PDF: {str(e)}")
```

```python
# USAGE EXAMPLE - Modify these paths for your files
input_file = "2024_nc_collection.pdf"  # Change this to your PDF path
output_file = "2024_nc_collection_no_footer.pdf"  # Change this to desired output path

# Optional: Check if input file exists
if os.path.exists(input_file):
    remove_footer_from_pdf(input_file, output_file, footer_height=65)
else:
    print(f"❌ Input file not found: {input_file}")
    print("Please update the 'input_file' variable with the correct path to your PDF")
```
